```python
import matplotlib.pyplot as plt
import numpy as np

# Daten definieren
categories = ['Required Specification\n(EAGLE-1 / OGS)', 'Prototype Performance\n(This Work)']
ranges = [(120.0, 1120.0), (83.5, 1120.5)]

# Figure und Achsen erstellen
fig, ax = plt.subplots(figsize=(9, 3.5), dpi=300)

# Farben und Stil
colors = ['#1f77b4', '#DE6FA1'] # Blau für Vorgabe, Grün für erreichtes Ziel
y_pos = [1, 0]

# Balken zeichnen (horizontaler Bereich von min bis max)
for i, (vmin, vmax) in enumerate(ranges):
    width = vmax - vmin
    ax.barh(y_pos[i], width, left=vmin, height=0.4, color=colors[i], alpha=0.85, edgecolor='black', linewidth=1)
    # Text innerhalb oder neben den Balken für exakten Wert
    ax.text((vmin + vmax) / 2, y_pos[i], f'{vmin} to {vmax} $\\mu$rad', 
            ha='center', va='center', color='white', fontweight='bold', fontsize=11)

# Achsenanpassungen
ax.set_yticks(y_pos)
ax.set_yticklabels(categories, fontsize=11, fontweight='bold')
ax.set_xlabel('Beam Divergence Half-Angle [$\\mu$rad]', fontsize=12, labelpad=10)
ax.set_title('Comparison of Required vs. Achieved Divergence Range', fontsize=13, pad=15, fontweight='bold')

# X-Achse skalieren, damit man den Bereich von 0 bis ca. 1200 sauber sieht
ax.set_xlim(0, 1200)

# Gitterlinien für bessere Lesbarkeit
ax.grid(axis='x', linestyle='--', alpha=0.5)

# Layout optimieren
plt.tight_layout()

# Als PDF speichern (ideal für LaTeX/Beamer)
plt.savefig('divergence_comparison_scale.pdf', format='pdf')
plt.show()
```
